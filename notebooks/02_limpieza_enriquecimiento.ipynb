```python
# Escribe tu código aquí para limpiar y enriquecer los datos
import pandas as pd
datos_crudos = pd.read_csv('../src/data/raw/housing/housing.csv')
# Son 5 datos no representa nada para el proyecto mi desición es eliminarlos
datos_crudos = datos_crudos[
    ~(datos_crudos['total_bedrooms'] > datos_crudos['total_rooms'])
]
```
